```python
# ALGORITMA KONVERSI CELSIUS KE FAHRENHEIT

# Input nilai Celsius
celsius = float(input("Masukkan suhu dalam Celsius: "))

# Menghitung konversi ke Fahrenheit
fahrenheit = (celsius * 9/5) + 32

# Menampilkan hasil
print(f"{celsius}°C sama dengan {fahrenheit}°F")
```
